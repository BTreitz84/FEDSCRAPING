# Monthly update -- scrape new H.4.1 releases and fold them into the clean series

Meant to be run roughly once a month. Three steps: where are we, scrape
whatever's new, then re-run the (already validated) build notebooks so
the new weeks flow into every output file. All logic lives in
`Helpers/h41_scrape_helpers.py` -- this notebook only calls it.

```python
import os, sys
sys.path.append(os.path.join(os.path.dirname(os.getcwd()), "Helpers"))
from h41_scrape_helpers import update_h41_archive, summarize_manifest, report_archive_status, rerun_notebook

BASE = os.path.dirname(os.getcwd())
HTML_DIR = os.path.join(BASE, "Data Input", "Raw Input", "h41_html")
NOTEBOOK_DIR = os.getcwd()
OUTPUT_PATHS = {
    "balance sheet": os.path.join(BASE, "Data Output", "H41_balance_sheet_clean_timeseries.xlsx"),
    "maturity": os.path.join(BASE, "Data Output", "H41_maturity_distribution_clean_timeseries.xlsx"),
    "regional": os.path.join(BASE, "Data Output", "H41_regional_clean_timeseries.xlsx"),
    "spliced (1914-2026)": os.path.join(BASE, "Data Output", "H41_balance_sheet_1914_2026_spliced.xlsx"),
}
```

## Step 1 -- where are we

```python
report_archive_status(HTML_DIR, OUTPUT_PATHS)
```

## Step 2 -- scrape whatever's new

Checks from 10 days before the most recent release on disk through today
-- the buffer covers a release landing a few days off its usual Thursday
slot. Already-downloaded files are never re-fetched.

```python
manifest = update_h41_archive(out_dir=HTML_DIR)
summarize_manifest(manifest)
```

## Step 3 -- attach: re-run the build notebooks

Re-running the full extraction (rather than a true incremental append) is
deliberate -- it's the same validated pipeline, it takes under 2 minutes
over the whole archive, and it can't drift from what the standalone
notebooks produce.

```python
rerun_notebook(os.path.join(NOTEBOOK_DIR, "H41_BALANCE_SHEET_CLEAN.ipynb"))
rerun_notebook(os.path.join(NOTEBOOK_DIR, "H41_MATURITY_CLEAN.ipynb"))
rerun_notebook(os.path.join(NOTEBOOK_DIR, "H41_REGIONAL_CLEAN.ipynb"))
rerun_notebook(os.path.join(NOTEBOOK_DIR, "JHU_1914_MATCH.ipynb"))
```

## Step 4 -- confirm the new state

```python
report_archive_status(HTML_DIR, OUTPUT_PATHS)
```
