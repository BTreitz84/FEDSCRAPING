# Fed Chair speeches/testimony -- scrape + save

Discovers every item (speech, testimony, statement) for each of the 18
configured Fed Board officials via the FRASER API, and downloads any PDF
not already present on disk. Idempotent: safe to re-run, only fetches what's
missing. Produces the raw link-level manifest (one row per PDF file).

```python
import os, sys
import pandas as pd
sys.path.append(os.path.join(os.path.dirname(os.getcwd()), "Helpers"))
import fed_speeches_helpers as fsh
from pathlib import Path

BASE = os.path.dirname(os.getcwd())
DOWNLOADS_DIR = Path(r"C:\Users\User\Desktop\BS ANALYSIS\FEDSCRAPING\speeches_chair\downloads")
MANIFEST_PATH = os.path.join(BASE, "Data Input", "Raw Input", "fed_speeches_manifest_raw.csv")
```

## Discover + download for every configured speaker

```python
manifests = []
for label, title_id in fsh.CHAIR_IDS.items():
    m = fsh.scrape_speaker(label, title_id, DOWNLOADS_DIR)
    manifests.append(m)
    print(f"{label} (title {title_id}): {len(m)} items, ok={m.attrs['dl_ok']} skip={m.attrs['dl_skip']} fail={m.attrs['dl_fail']}")

manifest = pd.concat(manifests, ignore_index=True)
print(f"\nTotal: {len(manifest):,} items across {manifest['chair_label'].nunique()} speakers")
```

## Save the raw manifest

```python
os.makedirs(os.path.dirname(MANIFEST_PATH), exist_ok=True)
manifest.to_csv(MANIFEST_PATH, index=False)
print(f"Saved {len(manifest):,} rows to: {MANIFEST_PATH}")
```
