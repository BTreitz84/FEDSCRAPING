# Fed Chair speeches/testimony -- compile

Takes the scraped manifest (one row per PDF file) and extracts RAW text
from every PDF with pdfplumber (no cleaning yet). This is the "raw, scraped
and compiled" dataset. Re-extracting with pdfplumber also fixes, as a side
effect, two pre-existing problems in the old dataset: mojibake-corrupted
text, and "=== PAGE BREAK ===" markers inserted by the old extraction
method (see PROGRESS_LOG.md).

```python
import os, sys
import pandas as pd
sys.path.append(os.path.join(os.path.dirname(os.getcwd()), "Helpers"))
import fed_speeches_helpers as fsh

BASE = os.path.dirname(os.getcwd())
MANIFEST_PATH = os.path.join(BASE, "Data Input", "Raw Input", "fed_speeches_manifest_raw.csv")
COMPILED_PATH = os.path.join(BASE, "Data Input", "Raw Input", "fed_speeches_compiled_raw.csv")
```

## Load the manifest

```python
manifest = pd.read_csv(MANIFEST_PATH)
print(manifest.shape)
manifest.head()
```

## Extract text from every PDF (this takes a while -- 2000+ files)

```python
compiled = fsh.compile_manifest_texts(manifest)
print(compiled.shape)
print(f"{compiled['error'].notna().sum()} extraction errors")
print(f"{compiled['needs_ocr'].sum()} scanned-image PDFs flagged needs_ocr (no text layer)")
compiled.head()
```

## Save the raw compiled dataset

```python
os.makedirs(os.path.dirname(COMPILED_PATH), exist_ok=True)
compiled.to_csv(COMPILED_PATH, index=False)
print(f"Saved {len(compiled):,} rows to: {COMPILED_PATH}")
```
