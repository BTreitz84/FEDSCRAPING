# FOMC statements -- scrape

Scrapes every historical FOMC document link (Statement/Minutes/
Transcripts/etc.) plus the recent/scheduled calendar, combines them,
and saves the raw link table. Fetching statement text happens in the
clean notebook, not here -- this step is just link discovery.

```python
import os, sys
sys.path.append(os.path.join(os.path.dirname(os.getcwd()), "Helpers"))
from fomc_helpers import scrape_all_fomc_docs, scrape_calendar, concat_historical_and_calendar

BASE = os.path.dirname(os.getcwd())
RAW_PATH = os.path.join(BASE, "Data Input", "Raw Input", "fomc_docs_raw.csv")
```

## Scrape historical document links

```python
df_hist = scrape_all_fomc_docs(pause=0.4, verbose=True)
print(df_hist.shape)
df_hist.head()
```

## Scrape the recent/scheduled calendar

```python
df_cal = scrape_calendar()
print(df_cal.shape)
df_cal.head()
```

## Combine and save

```python
df_combined = concat_historical_and_calendar(df_hist, df_cal)
print(df_combined.shape)

os.makedirs(os.path.dirname(RAW_PATH), exist_ok=True)
df_combined.to_csv(RAW_PATH, index=False)
print(f"Saved {len(df_combined):,} rows to: {RAW_PATH}")
```
