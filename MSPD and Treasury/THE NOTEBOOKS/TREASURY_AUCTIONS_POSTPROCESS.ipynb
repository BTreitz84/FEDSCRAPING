# Treasury Auctions -- produce the clean dataset

The raw fetch already type-coerces dates/numerics. This step drops rows
with no CUSIP (not a real security row) and saves the final deliverable
to Data Output.

```python
import os
import pandas as pd

BASE = os.path.dirname(os.getcwd())
RAW_PATH = os.path.join(BASE, "Data Input", "Raw Input", "treasury_auctions_raw.csv")
CLEAN_PATH = os.path.join(BASE, "Data Output", "treasury_auctions_clean.csv")
```

## Load raw, tidy, save

```python
auctions = pd.read_csv(RAW_PATH, parse_dates=["auction_date", "issue_date", "maturity_date", "announcemt_date"])
before = len(auctions)
auctions = auctions.dropna(subset=["cusip"]).drop_duplicates(subset=["cusip", "issue_date"]).sort_values("issue_date").reset_index(drop=True)
print(f"{len(auctions):,} / {before:,} rows kept after dropping no-CUSIP / duplicate rows")

os.makedirs(os.path.dirname(CLEAN_PATH), exist_ok=True)
auctions.to_csv(CLEAN_PATH, index=False)
print(f"Saved {len(auctions):,} rows to: {CLEAN_PATH}")
```
