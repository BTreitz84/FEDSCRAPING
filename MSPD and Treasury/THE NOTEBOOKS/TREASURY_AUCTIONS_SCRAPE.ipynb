# Treasury Auctions -- scrape full history

Pulls the complete Treasury auctions history from the FiscalData API
(`auctions_query`) and saves a raw snapshot. CSV, not Excel -- this
dataset is large enough (tens of thousands of rows across 46 years) that
CSV is the better fit.

```python
import os, sys
sys.path.append(os.path.join(os.path.dirname(os.getcwd()), "Helpers"))
from treasury_fiscaldata_helpers import fetch_auctions

BASE = os.path.dirname(os.getcwd())
RAW_PATH = os.path.join(BASE, "Data Input", "Raw Input", "treasury_auctions_raw.csv")
```

## Fetch full history

```python
auctions = fetch_auctions(start="1979-01-01", include_extra=True)
print(f"{len(auctions):,} rows, {auctions['issue_date'].min().date()} to {auctions['issue_date'].max().date()}")
auctions.head()
```

## Save raw snapshot

```python
os.makedirs(os.path.dirname(RAW_PATH), exist_ok=True)
auctions.to_csv(RAW_PATH, index=False)
print(f"Saved {len(auctions):,} rows to: {RAW_PATH}")
```
