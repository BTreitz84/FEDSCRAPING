# MSPD -- produce the clean datasets

Drops fully-empty columns (both tables carry many security-class columns
that are null for most rows) and saves the final deliverables.

```python
import os
import pandas as pd

BASE = os.path.dirname(os.getcwd())
RAW_PATH_3 = os.path.join(BASE, "Data Input", "Raw Input", "mspd_table3_raw.csv")
RAW_PATH_5 = os.path.join(BASE, "Data Input", "Raw Input", "mspd_table5_raw.csv")
CLEAN_PATH_3 = os.path.join(BASE, "Data Output", "mspd_table3_clean.csv")
CLEAN_PATH_5 = os.path.join(BASE, "Data Output", "mspd_table5_clean.csv")
```

## Load, drop empty columns, save

```python
mspd3 = pd.read_csv(RAW_PATH_3, parse_dates=["record_date"], low_memory=False)
mspd3 = mspd3.dropna(axis=1, how="all").drop_duplicates().sort_values("record_date").reset_index(drop=True)
print(f"table_3: {mspd3.shape[0]:,} rows x {mspd3.shape[1]} cols")

mspd5 = pd.read_csv(RAW_PATH_5, parse_dates=["record_date"], low_memory=False)
mspd5 = mspd5.dropna(axis=1, how="all").drop_duplicates().sort_values("record_date").reset_index(drop=True)
print(f"table_5: {mspd5.shape[0]:,} rows x {mspd5.shape[1]} cols")

os.makedirs(os.path.dirname(CLEAN_PATH_3), exist_ok=True)
mspd3.to_csv(CLEAN_PATH_3, index=False)
mspd5.to_csv(CLEAN_PATH_5, index=False)
print(f"Saved to: {CLEAN_PATH_3}")
print(f"Saved to: {CLEAN_PATH_5}")
```
