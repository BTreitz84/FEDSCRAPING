# MSPD (Monthly Statement of the Public Debt) -- scrape full history

Pulls `mspd_table_3` and `mspd_table_5` from the same FiscalData platform
Treasury Auctions uses. Both tables are large (the full `table_3` history
is 200k+ rows) -- CSV, not Excel.

```python
import os, sys
sys.path.append(os.path.join(os.path.dirname(os.getcwd()), "Helpers"))
from treasury_fiscaldata_helpers import fetch_mspd

BASE = os.path.dirname(os.getcwd())
RAW_PATH_3 = os.path.join(BASE, "Data Input", "Raw Input", "mspd_table3_raw.csv")
RAW_PATH_5 = os.path.join(BASE, "Data Input", "Raw Input", "mspd_table5_raw.csv")
```

## Fetch table_3 (detail by security)

```python
mspd3 = fetch_mspd(table="table_3", start="2001-01-01")
print(f"{len(mspd3):,} rows, {mspd3['record_date'].min().date()} to {mspd3['record_date'].max().date()}")
mspd3.head()
```

## Fetch table_5 (summary by category)

```python
mspd5 = fetch_mspd(table="table_5", start="2001-01-01")
print(f"{len(mspd5):,} rows, {mspd5['record_date'].min().date()} to {mspd5['record_date'].max().date()}")
mspd5.head()
```

## Save raw snapshots

```python
os.makedirs(os.path.dirname(RAW_PATH_3), exist_ok=True)
mspd3.to_csv(RAW_PATH_3, index=False)
mspd5.to_csv(RAW_PATH_5, index=False)
print(f"Saved {len(mspd3):,} rows to: {RAW_PATH_3}")
print(f"Saved {len(mspd5):,} rows to: {RAW_PATH_5}")
```
